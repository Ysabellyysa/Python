Step function
retorna 0 ou 1 utilizada para problemas linearmente separadas

```python
2 + 2
```

```python
def step_fuction(soma):
  if soma >= 1:
    return 1
  return 0
```

```python
step_fuction(-10)
```

```python
step_fuction(10)
```

Sigmoid
Utilizada para problemas de classificação com somente duas classes, retorna uma
probabilidade entre 0 e 1

```python
def sigmoid_fuction(soma):
  return 1 / (1 + np.exp(-soma))
```

```python
sigmoid_fuction(-10)
```

```python
sigmoid_fuction(10)
```

Hyperbolic
Retorna valores entre -1 e 0 e é utilizada em alguns problemas de visão computacional

```python
def tahn_fuction(soma):
  return(np.exp(soma) - np.exp(-soma)) / (np.exp(soma) + np.exp(-soma))
```

```python
tahn_fuction(-10)
```

```python
tahn_fuction(10)
```

```python
tahn_fuction(1)
```

ReLU
Retorna valores maiores do que 0 se o valor for negativo ele é zerado e é uma
função utilizada nas camadas ocutas das redes neurais.

```python
def relu_fanction(soma):
  if soma >= 0:
    return soma
  return 0
```

```python
relu_fanction(-20)
```

```python
relu_fanction(0)
```

```python
relu_fanction(10)
```

```python
relu_fanction(100)
```

Softmax
Retorna uma probabilidade para cada uma das classes é utilizada em problemas de classificalão com mais de duas classes.

```python
def softmax_function(x):
  ex = np.exp(x)
  return ex / ex.sum()
```

```python
softmax_function([7.0, 2.0, 1.3])
```

```python
softmax_function([2.0, 2.0, 5.3])
```
