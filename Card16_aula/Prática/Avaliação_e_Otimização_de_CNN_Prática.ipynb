```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

```python
# Simulando 64 imagens coloridas de 64x64 pixels para treino e 32 para teste
X_treino_simulado = np.random.rand(64, 64, 64, 3).astype('float32')
y_treino_simulado = np.random.randint(0, 2, size=(64, 1)).astype('float32')

X_teste_simulado = np.random.rand(32, 64, 64, 3).astype('float32')
y_teste_simulado = np.random.randint(0, 2, size=(32, 1)).astype('float32')
```

```python
gerador_treino = ImageDataGenerator(
    rescale=1./255,            # Normalização matemática de pixels (0-255 para 0-1)
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True
)
```

```python
gerador_teste = ImageDataGenerator(rescale=1./255)
```

```python
fluxo_treino = gerador_treino.flow(X_treino_simulado, y_treino_simulado, batch_size=8)
fluxo_teste = gerador_teste.flow(X_teste_simulado, y_teste_simulado, batch_size=8)
```

```python
classificador = Sequential()
# Aplicando 32 filtros com tamanho de matriz 3x3 e função de ativação ReLU
classificador.add(Conv2D(filters=32, kernel_size=(3, 3), input_shape=(64, 64, 3), activation='relu'))
# Reduzindo a amostragem espacial para capturar invariância e encolher a matriz
classificador.add(MaxPooling2D(pool_size=(2, 2)))
# Extração de padrões de maior ordem
classificador.add(Conv2D(filters=64, kernel_size=(3, 3), activation='relu'))
classificador.add(MaxPooling2D(pool_size=(2, 2)))
# Achatando a matriz bidimensional final em uma única coluna estruturada
classificador.add(Flatten())
# Conectando o vetor achatado a uma estrutura clássica de neurônios
classificador.add(Dense(units=128, activation='relu'))
classificador.add(Dropout(0.2)) # Camada de regularização contra overfitting
# Camada de Saída
classificador.add(Dense(units=1, activation='sigmoid'))
```

```python
classificador.compile( optimizer='adam', loss='binary_crossentropy',
    metrics=['accuracy'])
```

```python
historico = classificador.fit(fluxo_treino, epochs=5,validation_data=fluxo_teste)
```
