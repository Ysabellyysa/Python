```python
import numpy as np #importando bibliotecas 

your_word = "country" #palavra procurada

ms = nlp.vocab.vectors.most_similar(np.asarray( [nlp.vocab.vectors
                [nlp.vocab.strings[your_word]]]),n=10) 
#busca o indice que corresponde a palavra country no banco vetorial e pega as 10 palavras mais próximas 
words = [nlp.vocab.strings[w] for w in ms[0][0]]
distances=ms[2]
print(words)
```

```python
doc1 = nlp("I like salty fries and hamburgers.")
doc2 = nlp("Fast food tastes very good.")
# define dois docs
```

```python
print(doc1, "<->", doc2, doc1.similarity(doc2))
# ssa funcao calcula o quanto as dois docs são iguais semânticamente
```

```python
doc3 = nlp("The Empire State Building is in New York.")
```

```python
print(doc1, "<->", doc3, doc1.similarity(doc3))
#compara agora com um doc que nao possui semântica
```
