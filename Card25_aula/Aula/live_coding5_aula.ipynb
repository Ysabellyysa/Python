```python
from spacy.language import Language #importando 
```

```python
@Language.component("remove_gpe") #registra a função no pipeline
def remove_gpe(doc):
    original_ents = list(doc.ents)
#entidades originais do texto
    for ent in doc.ents:
#para cada entidade verifica se é geopolitica se for, vai remover do doc
        if ent.label_ == "GPE":
            original_ents.remove(ent)
    doc.ents = original_ents
    return (doc)
```

```python
nlp.add_pipe("remove_gpe") #adiciona ao pipeline
```

```python
nlp.analyze_pipes() #analisa com pipeline
```

```python
doc = nlp("Britain is a place. Mary is a doctor.")
for ent in doc.ents: #define um novo doc com o texto
    print (ent.text, ent.label_)
```
