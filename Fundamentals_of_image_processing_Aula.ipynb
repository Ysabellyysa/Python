### Reading Image using Opencv

```python
#importando bibliotecas
import cv2 as cv
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline
```

```python
img = cv.imread('69b82d004e87e434deb5bf0b04d0d10b.jpg') #leitura da imagem
```

### Opencv by defaults reads the image in BGR

```python
plt.imshow(img) #formato B0
```

```python
rgb_image = cv.cvtColor(img, cv.COLOR_BGR2RGB) #formato RGB
plt.imshow(rgb_image)
```

### Converting to Grayscale

```python
gray_image = cv.cvtColor(rgb_image, cv.COLOR_RGB2GRAY) #Convertendo a imagem RGB para uma imagem em tons de cinza
```

```python
plt.imshow(gray_image, cmap='gray')
```

```python
rgb_image.shape
```

```python
gray_image.shape
```

```python
img_gray_small = cv.resize(gray_image,(50,50)) #exibindo a imagem em tamanho menor
plt.imshow(img_gray_small, cmap='gray')
```

```python
img_gray_small.shape
```

```python
plt.imshow(gray_image[10:340,:], cmap='gray') #recorte de imagem selecionando uma região de interseção
```

```python
plt.imshow(gray_image[:,:300], cmap='gray')
```
