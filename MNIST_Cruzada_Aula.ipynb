```python
import keras
import numpy as np
import tensorflow as tf
import sklearn
```

```python
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import InputLayer, Dense, Flatten, Conv2D, MaxPooling2D
from tensorflow.keras import utils as np_utils
import numpy as np
from sklearn.model_selection import StratifiedKFold
```

```python
(X, y), (X_teste, y_teste) = mnist.load_data()
X = X.reshape(X.shape[0], 28, 28, 1)
X = X.astype('float32')
X /= 255
y = np_utils.to_categorical(y, 10)
```

```python
np.zeros(5)
```

```python
y.shape, y.shape[0]
```

```python
np.zeros(shape = (y.shape[0], 1)) #matriz de zeros com o mesmo número de linhas
```

```python
seed = 5
np.random.seed(seed) # Usando a seed para garantir os resultados
```

```python
kfold = StratifiedKFold(n_splits = 5, shuffle=True, random_state = seed) #memorizaçao da rede neural
```

```python
resultados = []
for indice_treinamento, indice_teste in kfold.split(X, np.zeros(shape = (y.shape[0], 1))):
  print("Índices treinamento: ", indice_treinamento, "Índice teste: ", indice_teste)
  classificador = Sequential()
  #Camada de convolução
  classificador.add(InputLayer(shape=(28, 28, 1)))
  classificador.add(Conv2D(32, (3, 3), activation='relu'))
  classificador.add(MaxPooling2D(pool_size=(2,2)))
  classificador.add(Flatten())
  # Camada oculta
  classificador.add(Dense(units=128, activation='relu'))
  # Camada saída
  classificador.add(Dense(units=10, activation='softmax'))
  classificador.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
  classificador.fit(X[indice_treinamento], y[indice_treinamento], batch_size = 128, epochs = 5) #usando dados de treino com o índice de treinamento
  precisao = classificador.evaluate(X[indice_teste], y[indice_teste]) # usando dados de teste com o índice de teste
  resultados.append(precisao[1])
```

```python
resultados
```

```python
results = np.array(resultados)
```
