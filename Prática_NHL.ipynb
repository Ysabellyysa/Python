```python
#Importando bibliotecas
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense, Dropout
from sklearn.metrics import confusion_matrix, accuracy_score, mean_absolute_error
```

```python
df = pd.read_csv('Hóquei.csv.csv') #carregamento de dados
df
```

```python
# Definição de Atributos (X) e Alvos (y)
X = df.drop(columns=['won', 'goals', 'pim']).values
y_class = df['won'].values   # Alvo Binário
y_reg_goals = df['goals'].values # Alvo Regressão 1
y_reg_pim = df['pim'].values   # Alvo Regressão 2
```

```python
# Treino, teste e Escalonamento
X_train, X_test, y_c_train, y_c_test, y_g_train, y_g_test, y_p_train, y_p_test = train_test_split(
    X, y_class, y_reg_goals, y_reg_pim, test_size=0.2, random_state=42
)
```

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
input_layer = Input(shape=(X.shape[1],)) #arquitetura de rede
```

```python
#Camadas Ocultas
shared = Dense(64, activation='relu')(input_layer)
shared = Dropout(0.3)(shared) # Prevenção de Overfitting
shared = Dense(32, activation='relu')(shared)
```

```python
out_class = Dense(1, activation='sigmoid', name='vitoria')(shared)
out_goals = Dense(1, activation='linear', name='gols')(shared)
out_pim = Dense(1, activation='linear', name='penalidades')(shared)
```

```python
#Montagem do Modelo
model = Model(inputs=input_layer, outputs=[out_class, out_goals, out_pim])
```

```python
#copilação e treinamento
model.compile(
    optimizer='adam',
    loss={'vitoria': 'binary_crossentropy', 'gols': 'mse', 'penalidades': 'mse'},
    metrics={'vitoria': 'accuracy', 'gols': 'mae', 'penalidades': 'mae'}
)
```

```python
model.fit(X_train, [y_c_train, y_g_train, y_p_train], epochs=50, batch_size=32, verbose=1)
```

```python
previsoes = model.predict(X_test) #previsões do modelo para os dados de teste
```

```python
# transformação de probabilidade em 0 ou 1
vitoria_pred = (previsoes[0] > 0.5).astype(int)
vitoria_real = y_c_test.flatten()
```

```python
#Comparação de quantos são iguais
acertos_vitoria = (vitoria_pred == vitoria_real).sum()
total_jogos = len(vitoria_real)
porcentagem_acerto = (acertos_vitoria / total_jogos) * 100
```
