```python
import torch
```

```python
#Criação de tensor 1D
tensor_1d = torch.tensor([25, 30, 35])
tensor_1d.shape
```

```python
#Criação de tensor 2D
tensor_2d = torch.tensor([[5, 15],
                          [20, 25],
                          [30, 35]])
tensor_2d.shape
```

```python
#Criação do tensor 3D
tensor_3d = torch.tensor([[[2, 4], [6, 8]],
                          [[10, 12], [14, 16]]])
tensor_3d.shape
```

```python
#Multiplica os elementos por 5
multi_escalar = tensor_2d*5
multi_escalar
```

```python
#Soma dos elementos
soma_tensor = tensor_2d.add(5)
soma_tensor
```

```python
#visualização do tensor 2d
tensor_2d
```

```python
#Remodelagem
mold_tensor = tensor_2d.view(2,3)
mold_tensor
```

```python
#criando um tensor com dimensões para testar o squeeze
dim_tensor = torch.ones(1, 3, 1)
dim_tensor
```

```python
#Removendo dimensões de tamanho 1
tam_tensor = dim_tensor.squeeze(0).squeeze(1)
tam_tensor.shape
```

```python
unsquze_tensor = tam_tensor.unsqueeze(1)
unsquze_tensor.shape
tam_tensor[:, None].shape
```

```python
#Definindo matriz A de dimensões e matriz B de dimensões
matriz_a = torch.tensor([[1, 2],
                         [3, 4],
                         [5, 6]])

matriz_b = torch.tensor([[10, 20, 30, 40],
                         [50, 60, 70, 80]])
```

```python
#Multiplicação de matriz
matriz_result = torch.matmul(matriz_a, matriz_b)
matriz_result
matriz_result.shape
```

```python
#Simulando imagem
imagem_tensor = torch.rand(3, 15, 10)
imagem_tensor.shape
```

```python
#alteração de ordem dos eixos
eixo_tensor = imagem_tensor.permute(1, 2, 0)
eixo_tensor.shape
```
