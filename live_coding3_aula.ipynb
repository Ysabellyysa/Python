```python
import spacy
```

```python
nlp = spacy.load("en_core_web_sm")
text = "West Chestertenfieldville was referenced in Mr. Deeds."
```

```python
doc = nlp(text)
# cria um doc com o texto anterior
```

```python
for ent in doc.ents:
    print (ent.text, ent.label_)
#entidades encontradas no texto baseado em machine learning
```

```python
ruler = nlp.add_pipe("entity_ruler")
# adiciona ao pipeline
```

```python
nlp.analyze_pipes()
# mostrando o  pipeline
```

```python
patterns = [ {"label": "GPE", "pattern": "West Chestertenfieldville"}]
```

```python
ruler.add_patterns(patterns)
```

```python
doc2 = nlp(text)
for ent in doc2.ents:
    print (ent.text, ent.label_)
```

```python
nlp2 = spacy.load("en_core_web_sm")
# cria um pipeline sem nada
```

```python
ruler = nlp2.add_pipe("entity_ruler", before="ner")
# adiciona a regra antes do NER
```

```python
ruler.add_patterns(patterns)
```

```python
doc = nlp2(text)
```

```python
for ent in doc.ents:
    print (ent.text, ent.label_)
    # agora ao aplicar a regra podemos ver que mudou 
```

```python
nlp2.analyze_pipes()
```
